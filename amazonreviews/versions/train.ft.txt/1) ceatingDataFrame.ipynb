<h2>Import needed libraries</h2>

```python
import pandas as pd
```

<h3>Read the data line by line</h3>

```python
with open("train.ft.txt" , "r" , encoding="utf-8") as file :
    listOfLines = file.readlines()
```

<h3>Prepare the DataFrame</h3>
<small>The data follows the following format :<br>__label__&lt;X&gt; __label__&lt;Y&gt; ... &lt;text&gt; <br>where X and Y are the class names. No quotes, all on one line.<br>
In this case, the classes are __label__1 and __label__2, and there is only one class per row.<br>
__label__1 corresponds to 1- and 2-star reviews, and __label__2 corresponds to 4- and 5-star reviews.<br>
(3-star reviews i.e. reviews with neutral sentiment were not included in the original),<br>
The review titles, followed by ':' and a space, are prepended to the text.<br>
Most of the reviews are in English, but there are a few in other languages, like Spanish.</small>

```python
listOfRatings = []
listOfReviews = []

for line in listOfLines:
    line = line.replace('\n' , '')
    for rating in line.split(" " , 1)[0][-1]:
        listOfRatings.append(rating)
    listOfReviews.append(line.split(" " , 1)[1])
```

```python
df = pd.DataFrame({"Review" : listOfReviews ,
                   "Rating" : listOfRatings})

df["Review"] = df["Review"].astype("object")
df["Rating"] = df["Rating"].astype("int32")

df.to_csv("trainDataFrame.csv")
```
